```python
import pandas as pd
from pycaret.regression import *
```

```python
df = pd.read_csv('datasets/dataset_with_targets.csv')
#df.head()
```

```python
# Define your five target variables as a list of strings
# List of regression targets
targets = ['monsoon_intensity', 
           'climate_change', 
           'siltation',
           'agricultural_practices', 
           'landslide_risks'
        ]

# A dictionary to store the best model for each target
trained_models = {}
```

```python
for target in targets:
    print("-" * 50)
    print(f"🚀 Training models for target: '{target}'")
    print("-" * 50)

    # Prepare data
    features = df.drop(columns=['city', 'date'] + targets)
    data = df[features.columns.tolist() + [target]]

    # Setup
    setup(data=data, target=target, session_id=123, verbose=True)

    # Train and compare models
    best_model = compare_models(exclude=['lightgbm'],verbose=False)

    # Evaluate model
    print(f"\n📈 Evaluating best model for: {target}")
    evaluate_model(best_model)


    # Finalize and save model
    final_model = finalize_model(best_model)
    save_model(final_model, f"best_model_{target}")

    # Store reference
    trained_models[target] = final_model

print("\n✅ All models trained, evaluated, and saved successfully.")
```
