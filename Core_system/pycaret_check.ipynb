```python


import pandas as pd
from pycaret.regression import *
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
```

```python
flood_data = pd.read_csv('data/flood.csv')
```

```python
numerical_features = ['year', 'rainfall','runoff']
categorical_features = ['month', 'region']

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='mean')),
    #('scaler', StandardScaler())
])

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numerical_features),
        ('cat_transform', OneHotEncoder(handle_unknown='ignore', sparse_output=False), categorical_features)
        ],
    remainder='passthrough'
)
```

```python
flood_data.head()
```

```python
s = setup(data=flood_data, target='FloodProbability', session_id=5)
```

```python
best = compare_models(sort='MAE', exclude=['lightgbm'])
results = pull()
print(results.iloc[0])
```
