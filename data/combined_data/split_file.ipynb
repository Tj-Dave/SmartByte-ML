```python
import pandas as pd

# Define the current year based on the context provided in the prompt
current_year = 2025

# Load the transformed dataset
df_transformed = pd.read_csv('lakevic_kagera_simiyu_victorianile_transformed.csv')
print(df_transformed.head())

# Split the data into historical and projected based on the 'year' column
historical_data = df_transformed[df_transformed['year'] <= current_year]
projected_data = df_transformed[df_transformed['year'] > current_year]

# Define output file names
historical_output_file = 'historical_lakevic_data.csv'
projected_output_file = 'projected_lakevic_data.csv'

# Save the historical data to a CSV file
historical_data.to_csv(historical_output_file, index=False)

# Save the projected data to a CSV file
projected_data.to_csv(projected_output_file, index=False)
```
